```python
# ==========================================
# Data Visualization using Matplotlib & Seaborn
# ==========================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set plotting style
plt.style.use("ggplot")

print("Libraries Imported Successfully")
```

```python
# Load Titanic dataset from GitHub

url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"

df = pd.read_csv(url)

print("Dataset Loaded Successfully")

df.head()
```

```python
print("Dataset Information")
print("="*50)

df.info()
```

```python
print("Statistical Summary")
print("="*50)

df.describe()
```

```python
print("Missing Values")
print("="*50)

df.isnull().sum()
```

```python
print("Five Observations")
print("="*50)

print("1. Dataset contains", df.shape[0], "rows and", df.shape[1], "columns.")

print("2. Age column contains missing values.")

print("3. Cabin has a large number of missing values.")

print("4. Dataset contains both numerical and categorical columns.")

print("5. PassengerId is unique for every passenger.")
```

```python
numeric_cols = df.select_dtypes(include=['int64','float64']).columns

for col in numeric_cols:

    plt.figure(figsize=(6,4))

    plt.hist(df[col].dropna(), bins=30)

    plt.title(f"Distribution of {col}")

    plt.xlabel(col)

    plt.ylabel("Frequency")

    plt.show()
```

```python
for col in numeric_cols:

    plt.figure(figsize=(6,4))

    sns.histplot(df[col], kde=True)

    plt.title(f"Seaborn Distribution of {col}")

    plt.show()
```

```python
for col in numeric_cols:

    plt.figure(figsize=(6,4))

    sns.boxplot(x=df[col])

    plt.title(f"Boxplot of {col}")

    plt.show()
```

```python
plt.figure(figsize=(10,8))

sns.heatmap(df[numeric_cols].corr(),
            annot=True,
            cmap="coolwarm",
            fmt=".2f")

plt.title("Correlation Heatmap")

plt.show()
```

```python
cat_cols = df.select_dtypes(include='object').columns

for col in cat_cols:

    plt.figure(figsize=(8,5))

    df[col].value_counts().head(10).plot(kind="bar")

    plt.title(f"Top 10 Categories - {col}")

    plt.xlabel(col)

    plt.ylabel("Count")

    plt.xticks(rotation=45)

    plt.show()
```

```python
plt.figure(figsize=(6,4))

sns.countplot(data=df, x="Survived")

plt.title("Passengers Survival Count")

plt.show()
```

```python
plt.figure(figsize=(6,4))

sns.countplot(data=df,
              x="Sex",
              hue="Survived")

plt.title("Survival by Gender")

plt.show()
```

```python
plt.figure(figsize=(6,4))

sns.countplot(data=df,
              x="Pclass",
              hue="Survived")

plt.title("Passenger Class vs Survival")

plt.show()
```

```python
plt.figure(figsize=(7,5))

sns.scatterplot(data=df,
                x="Age",
                y="Fare",
                hue="Survived")

plt.title("Age vs Fare")

plt.show()
```

```python
sns.pairplot(df[['Age','Fare','Pclass','Survived']])

plt.show()
```
