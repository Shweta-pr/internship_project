```python
import numpy as np

# 1D Array
arr1 = np.array([10, 20, 30, 40, 50])

# 2D Array
arr2 = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

# 3D Array
arr3 = np.array([
    [
        [1, 2],
        [3, 4]
    ],
    [
        [5, 6],
        [7, 8]
    ]
])

print("1D Array:")
print(arr1)

print("\n2D Array:")
print(arr2)

print("\n3D Array:")
print(arr3)
```

```python
print("Shape of 1D Array:", arr1.shape)
print("Shape of 2D Array:", arr2.shape)
print("Shape of 3D Array:", arr3.shape)
```

```python
matrix = np.array([
    [1,2,3],
    [4,5,6]
])

vector = np.array([10,20,30])

result = matrix + vector

print("Matrix:")
print(matrix)

print("\nVector:")
print(vector)

print("\nBroadcasting Result:")
print(result)
```

```python
a = np.array([1,2,3,4,5])
b = np.array([6,7,8,9,10])

print("Addition:")
print(a + b)

print("Subtraction:")
print(a - b)

print("Multiplication:")
print(a * b)

print("Division:")
print(a / b)

print("Square:")
print(a ** 2)
```

```python
A = np.array([
    [1,2],
    [3,4]
])

B = np.array([
    [5,6],
    [7,8]
])

C = A @ B

print("Matrix A")
print(A)

print("\nMatrix B")
print(B)

print("\nMatrix Multiplication")
print(C)
```

```python
import pandas as pd

url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/iris.csv"

df = pd.read_csv(url)

df.head()
```

```python
numeric_data = df.select_dtypes(include='number')

data = numeric_data.to_numpy()

print(data[:5])
```

```python
mean = np.mean(data, axis=0)

print("Mean")
print(mean)
```

```python
std = np.std(data, axis=0)

print("Standard Deviation")
print(std)
```

```python
correlation = np.corrcoef(data, rowvar=False)

print("Correlation Matrix")
print(correlation)
```

```python
columns = numeric_data.columns

print("Mean Values")
for c, m in zip(columns, mean):
    print(f"{c}: {m:.2f}")

print("\nStandard Deviation")
for c, s in zip(columns, std):
    print(f"{c}: {s:.2f}")

print("\nCorrelation Matrix")
print(correlation)
```
